```python
#   Valor calórico
#  Se imprimen dos boxplots con y sin outliers 
with pd.option_context('display.max_rows', None):
    print(dataset["energy_kcal"].value_counts(dropna=False))

sns.boxplot(dataset["energy_kcal"])
plt.title("Boxplot del valor calórico")
plt.ylabel("kCal")

plt.show()

sns.boxplot(dataset[dataset["energy_kcal"]<1000]["energy_kcal"])
plt.title("Boxplot del valor calórico (Sin outliers)")
plt.ylabel("kCal")

plt.show()
```

```python
#   Grasas en 100g
with pd.option_context('display.max_rows', None):
    print(dataset["fat_100g"].sort_index().value_counts(dropna=False))

mayores_100 = dataset[dataset["fat_100g"]>100]
mayores_100.head(10)
```

```python
#   Grasas en 100g

with pd.option_context('display.max_rows', None):
    print(dataset["fat_100g"].value_counts(dropna=False))

sns.boxplot(dataset["fat_100g"])
plt.title("Boxplot las grasas en 100g")


plt.show()

sns.boxplot(dataset[dataset["fat_100g"]<=100]["fat_100g"])
plt.title("Boxplot las grasas en 100g (Sin mayores a 100)")

plt.show()
```

```python
#   Grasas saturadas en 100g
with pd.option_context('display.max_rows', None):
    print(dataset["saturated_fat_100g"].value_counts(dropna=False))

sns.boxplot(dataset["saturated_fat_100g"])
plt.title("Grasas saturadas en 100g")
plt.show()

sns.boxplot(dataset[dataset["saturated_fat_100g"]<=100]["saturated_fat_100g"])
plt.title("Grasas saturadas en 100g (Sin valores mayores a 100g)")
plt.show()

sns.boxplot(dataset[dataset["saturated_fat_100g"]<=50]["saturated_fat_100g"])
plt.title("Grasas saturadas en 100g (Sin valores mayores a 50g)")
plt.show()
```
