## Hipótesis 1

##### ¿Más del 50% de los productos ofrecidos son de mala calidad (un nutriscore con valor D o E)?

#### 1.1 - Limpieza nutriscore
Vamos a limpiar la columna nutriscore. No vamos a considerar los productos donde no aplica esta clasificación y vamos a traducir los unknown al tipo nulo de pandas.

```python
nutriscore_inicial = numericos[numericos["nutriscore_grade"]!="NOT-APPLICABLE"]
nutriscore_inicial["nutriscore_grade"] = nutriscore_inicial["nutriscore_grade"].replace("UNKNOWN",pd.NA)
print("Nuevo tamaño",len(nutriscore_inicial))
print("Cantidad de nulos",nutriscore_inicial["nutriscore_grade"].isna().sum())
```

```python
nutriscore_inicial["nutriscore_grade"] = nutriscore_inicial["nutriscore_grade"].map({"A": 1, "B": 2, "C": 3, "D": 4, "E": 5})

nutriscore_num = nutriscore_inicial.select_dtypes(include=np.number)
```

```python
#   Cuantos nulos hay cuando nova_group es nulo?
def graficar_nulos(dataframe, col, title=""):
    nulos = dataframe[dataframe[col].isna()]
    cant_nulos = nulos.isna().sum(axis=1)
    frecuencia = cant_nulos.value_counts().sort_index()
    ax = frecuencia.plot(kind="bar", figsize=(8, 5), edgecolor="black")
    ax.set_xlabel("Cantidad de nulos en la fila cuando col es nulo")
    ax.set_ylabel("Cantidad de filas")
    title +=" - Columnas totales:"+str(len(nulos.columns))
    ax.set_title(title)

    # Etiquetas con el valor sobre cada barra
    ax.bar_label(ax.containers[0])

    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
graficar_nulos(nutriscore_num,"nutriscore_grade","Frecuencia de nulos en columnas numéricas")
```

Para la imputación de valores consideramos que si una fila tiene demasiados nulos, no hay la información suficiente para asegurarse de que esta se está haciendo correctamente. Por eso es que para este análisis vamos a eliminar todas las filas donde hay 5 o más nulos. 

```python
N = 5
print("Columnas con nutriscore nulo: ",nutriscore_num["nutriscore_grade"].isna().sum())
nutriscore_num_nulos = nutriscore_num[(nutriscore_num["nutriscore_grade"].isna()) & (nutriscore_num.isna().sum(axis=1) < N)]
print("Columnas con nutriscore nulo y menos de 5 nulos: ",len(nutriscore_num_nulos))
graficar_nulos(nutriscore_num_nulos,"nutriscore_grade","Frecuencia de nulos en filas con menos de 5 nulos")
```

```python
from sklearn.impute import KNNImputer
from sklearn.preprocessing import MinMaxScaler

imputer = KNNImputer(n_neighbors=31 , metric="nan_euclidean")

filtro_muchos_nulos = nutriscore_inicial["nutriscore_grade"].isna() & (nutriscore_inicial.isna().sum(axis=1) >= N)
nutriscore_inicial = nutriscore_inicial[~filtro_muchos_nulos]
print(len(nutriscore_inicial))
nutriscore_limpio_num = nutriscore_num.copy()
nutriscore_limpio_num["nutriscore_grade"] = nutriscore_inicial["nutriscore_grade"]
scaler = MinMaxScaler()
dataset_stand = scaler.fit_transform(nutriscore_limpio_num)
matriz_imputada = imputer.fit_transform(dataset_stand)


df_imputado = pd.DataFrame(
    matriz_imputada,
    columns=nutriscore_limpio_num.columns,
    index=nutriscore_limpio_num.index
)


datos_imputados = scaler.inverse_transform(KNNImputer(n_neighbors=31).fit_transform(df_imputado))
df_imputados = pd.DataFrame(datos_imputados, columns=nutriscore_limpio_num.columns, index=nutriscore_limpio_num.index)
df_imputados.describe()
nutriscore_limpio = nutriscore_inicial.copy()
nutriscore_limpio["nutriscore_grade"] = df_imputados["nutriscore_grade"].apply(lambda x: "A" if x<=1.5 else "B" if x<=2.5 else "C" if x<=3.5 else "D" if x<4.5 else "E")

nutriscore_limpio["nutriscore_grade"].value_counts()
```

```python
dataset_viejo = dataset[dataset["nutriscore_grade"].isin(["A","B","C","D","E"])]

comparacion = pd.DataFrame({
    "Nuevos valores": nutriscore_limpio["nutriscore_grade"].value_counts().sort_index(),
    "Dataset original": dataset_viejo["nutriscore_grade"].value_counts().sort_index(),
}).fillna(0)

ax = comparacion.plot(kind="bar", edgecolor="black", figsize=(8, 5))
#ax.set_xlabel(col)
ax.set_ylabel("Frecuencia")
ax.set_title("Comparación entre frecuencias de nutriscore antes y después de imputar")
plt.xticks(rotation=0)

for c in ax.containers:
    ax.bar_label(c)

plt.tight_layout()
plt.show()
```

#### 1.2 - Análisis de la hipótesis

Para verificar la hipótesis, se debe hacer el Test de Hipótesis para una Proporción Poblacional (Prueba Z de una Proporción). 

```python
from statsmodels.stats.proportion import proportions_ztest

# Crear columna es_malacalidad: es 1 si es mala calidad segun nutriscore (D o E)
h1_dataset = nutriscore_limpio.copy()
h1_dataset['es_malacalidad'] = h1_dataset['nutriscore_grade'].isin(['D', 'E']).astype(int)

# Métricas de la muestra
n = h1_dataset['es_malacalidad'].count()         # Número total de observaciones
x = h1_dataset['es_malacalidad'].sum()           # Cantidad con es_malacalidad = 1
p_muest = x / n                            # Proporción muestral
p_hip = 0.50                                # Valor de la hipótesis nula

# Ejecutar la prueba de hipótesis para proporciones
z_stat, p_value = proportions_ztest(count=x, nobs=n, value=p_hip, alternative='larger')

# Mostrar resultados
print("=== RESULTADOS DEL TEST DE HIPÓTESIS ===")
print(f"Tamaño muestral (n)            : {n}")
print(f"Productos de mala calidad (x)  : {x}")
print(f"Proporción observada (p̂)       : {p_muest:.4f} ({p_muest*100:.2f}%)")
print(f"Estadístico Z de prueba        : {z_stat:.4f}")
print(f"p-valor (p-value)              : {p_value:.4f}")

# Conclusión estadística
alpha = 0.05
if p_value < alpha:
    print(f"\nDecisión: Rechazar H0 (p-valor < {alpha}).")
    print("Conclusión: Existe evidencia suficiente para afirmar que más del 50% de los productos son de mala calidad.")
else:
    print(f"\nDecisión: No rechazar H0 (p-valor >= {alpha}).")
    print("Conclusión: No existe evidencia suficiente para afirmar que más del 50% de los productos son de mala calidad.")


```

## Hipótesis 2
##### ¿Los productos con mayor grado de procesamiento industrial (nova_group)  son los que más impacto ambiental (ecoscore_grade) generan?

#### 2.1 - Análisis de la hipótesis

Para el análisis de la hipótesis, decidimos realizar el test de la correlación de Spearman. Como paso complementario, realizamos primero un heatmap para ver la distribución de valores entre ambas variables:

```python
h2_dataset = dataset.copy()

h2_dataset['ecoscore_grade'] = h2_dataset['ecoscore_grade'].replace(['-1', 'UNKNOWN', 'NOT-APPLICABLE'], pd.NA)

print(h2_dataset['ecoscore_grade'].isna().sum())

h2_dataset = h2_dataset.dropna(subset=['ecoscore_grade', 'nova_group'])

print(h2_dataset['ecoscore_grade'].isna().sum())

ct = pd.crosstab(h2_dataset['nova_group'], h2_dataset['ecoscore_grade'], normalize='columns') # normaliza para ver % por columna

plt.figure(figsize=(8, 6))
sns.heatmap(ct, annot=True, cmap="YlGnBu", fmt=".2f")
plt.title("Proporción de Nova Group por grado de Ecoscore")
plt.xlabel("Ecoscore")
plt.ylabel("Nova Group")
plt.gca().invert_yaxis() 
plt.show()
```

A simple vista, se puede ver una relación muy ligera entre el aumento del impacto ambiental con el aumento del procesamiento industrial, ya que vemos una leve concentración de valores en la esquina superior derecha.  
Para demostrar definitivamente esta hipótesis, debemos elegir un test que funcione en muestras no apareadas (ya que no consideramos que haya una dependencia entre los valores del ecoscore con el nova group). Como son variables de pocas categorías y ordinales, utilizaremos el test no paramétrico de Spearman, que mide la fuerza y el sentido de la relación (si al aumentar nova_group tiende a aumentar el ecoscore_grade).

```python
import scipy.stats as stats

# mapeo de datos
mapeo_eco = {'A-PLUS':0, 'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5}
h2_dataset['ecoscore_numerico'] = h2_dataset['ecoscore_grade'].map(mapeo_eco)
h2_dataset = h2_dataset.dropna(subset=['ecoscore_numerico', 'nova_group']) 

# ejecutamos la correlación
coef_corr, p_val_spearman = stats.spearmanr(h2_dataset['nova_group'], h2_dataset['ecoscore_numerico'])
print(f"Coeficiente de correlación de Spearman: {coef_corr:.4f}")
print(f"p-valor: {p_val_spearman}")

# toma de decisiones (usando alfa = 0.05)
print("\n--- CONCLUSIÓN ---")
if p_val_spearman < 0.05:
    print("Rechazamos la hipótesis nula (H0). Existe una relación estadísticamente significativa.")
    if coef_corr > 0:
        print("La correlación es positiva: a mayor grado de procesamiento (NOVA), mayor es el número de EcoScore (lo que significa peor impacto ambiental, ya que E=5).")
    else:
        print("La correlación es inversa.")
else:
    print("No se puede rechazar la hipótesis nula (H0). No hay evidencia estadística de que el nivel de procesamiento afecte al impacto ambiental en esta muestra.")
```

Podemos concluir entonces que, como el p-valor es menor a nuestro alfa (0.05), existe una relación estadísticamente significativa entre el grado de procesamientoy el impacto ambiental de un producto. Esta correlación es positiva, es decir, a medida que crece el grado de procesamiento, crece el impacto ambiental, y viceversa.

Imputación por k vecinos:

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import KNNImputer

h3_completo = h3_dataset.copy()
cols = ['nova_group', 'sugars_100g', 'saturated_fat_100g', 'energy_kcal']
h3_knn = h3_completo.copy()

# Valores imposibles -> NaN (para que no distorsionen el escalado ni los vecinos)
h3_knn.loc[~h3_knn['energy_kcal'].between(0, 900), 'energy_kcal'] = np.nan
h3_knn.loc[~h3_knn['sugars_100g'].between(0, 100), 'sugars_100g'] = np.nan
h3_knn.loc[~h3_knn['saturated_fat_100g'].between(0, 100), 'saturated_fat_100g'] = np.nan
h3_knn.loc[~h3_knn['nova_group'].isin([1, 2, 3, 4]), 'nova_group'] = np.nan

# Registro de qué valores eran nulos (para comparar después)
nulos_antes = h3_knn[cols].isna()
print("Nulos antes de imputar:\n", nulos_antes.sum())

# Escalar a [0, 1]: KNN usa distancias, así que sin escalar dominaría energy_kcal
# (MinMaxScaler ignora los NaN al ajustar y los conserva al transformar)
scaler = MinMaxScaler()
X = scaler.fit_transform(h3_knn[cols])

# Imputación por k vecinos
imputer = KNNImputer(n_neighbors=5, weights='distance')
X_imp = imputer.fit_transform(X)

# Volver a la escala original
h3_knn[cols] = scaler.inverse_transform(X_imp)

# nova_group es categórica ordinal: redondear a un grado válido (1 a 4)
h3_knn['nova_group'] = h3_knn['nova_group'].round().clip(1, 4).astype(int)

# volvemos a calcular nova_alto para los valores imputados
h3_knn["nova_alto"]= h3_knn["nova_group"].apply(lambda x: 1 if (x==4 or x==3) else 0)

# Chequeos
print("\nNulos después de imputar:\n", h3_knn[cols].isna().sum())
```

```python
import seaborn as sns
from scipy.stats import ks_2samp

antes, despues = h3_completo, h3_knn
c_antes, c_desp = '#1f77b4', '#d62728'

rangos = {'sugars_100g': (0, 100),
          'saturated_fat_100g': (0, 100),
          'energy_kcal': (0, 900)}

# Grilla de 2 filas x 4 columnas (la primera columna es para nova_group)
fig = plt.figure(figsize=(21, 9))
gs = fig.add_gridspec(2, 4, width_ratios=[0.9, 1, 1, 1])

# ------------------ nova_group (discreta): una barra por grado ------------------
ax = fig.add_subplot(gs[:, 0])            # ocupa las dos filas
niveles = [1, 2, 3, 4]
p_a = antes['nova_group'].value_counts(normalize=True).reindex(niveles, fill_value=0)
p_d = despues['nova_group'].value_counts(normalize=True).reindex(niveles, fill_value=0)
x = np.arange(len(niveles)); w = 0.4
ax.bar(x - w/2, p_a.values, w, color=c_antes, edgecolor='black', label='Antes')
ax.bar(x + w/2, p_d.values, w, color=c_desp,  edgecolor='black', label='Después (KNN)')
ax.set_xticks(x); ax.set_xticklabels([f'NOVA {n}' for n in niveles])
ax.set_title('nova_group', fontsize=12, fontweight='bold')
ax.set_ylabel('Proporción')
ax.legend(); ax.grid(True, axis='y', linestyle=':', alpha=0.6)

# ------------------ Variables continuas: KDE (arriba) y ECDF (abajo) ------------------
for j, (col, (lo, hi)) in enumerate(rangos.items(), start=1):
    a = antes[col].dropna()
    a = a[(a >= lo) & (a <= hi)]
    d = despues[col].dropna()
    d = d[(d >= lo) & (d <= hi)]

    # KS: 0 = distribuciones idénticas, 1 = totalmente distintas
    ks, p = ks_2samp(a, d)

    # Fila 1: densidad (KDE), recortada al rango válido
    ax = fig.add_subplot(gs[0, j])
    sns.kdeplot(a, ax=ax, color=c_antes, fill=True, alpha=0.35, linewidth=2,
                cut=0, clip=(lo, hi), label=f'Antes (n={len(a):,})')
    sns.kdeplot(d, ax=ax, color=c_desp, fill=True, alpha=0.35, linewidth=2,
                cut=0, clip=(lo, hi), label=f'Después (n={len(d):,})')
    ax.axvline(a.median(), color=c_antes, linestyle='--', linewidth=1.5)
    ax.axvline(d.median(), color=c_desp, linestyle='--', linewidth=1.5)
    ax.set_title(f'{col}\nKS = {ks:.3f}', fontsize=12, fontweight='bold')
    ax.set_xlabel(col); ax.set_ylabel('Densidad')
    ax.legend(); ax.grid(True, linestyle=':', alpha=0.6)

    # Fila 2: ECDF (muestra saltos y acumulaciones en un valor)
    ax = fig.add_subplot(gs[1, j])
    sns.ecdfplot(a, ax=ax, color=c_antes, linewidth=2, label='Antes')
    sns.ecdfplot(d, ax=ax, color=c_desp, linewidth=2, linestyle='--', label='Después')
    ax.set_xlabel(col); ax.set_ylabel('Proporción acumulada')
    ax.legend(); ax.grid(True, linestyle=':', alpha=0.6)

fig.suptitle('Distribución antes y después de la imputación por KNN',
             fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```

Podemos observar que no cambió de forma significativa la distribución de los valores.

##### 3.2.1 - Análisis con dataset imputado por mediana

```python
import statsmodels.api as sm
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score
from statsmodels.stats.outliers_influence import variance_inflation_factor

df = h3_impt.copy()   
features = ['energy_kcal', 'sugars_100g', 'saturated_fat_100g']
y_col = 'nova_alto'    

df = df[features + [y_col]].dropna()
y = df[y_col].astype(int)

# Estandarizar para que los coeficientes sean comparables
X = pd.DataFrame(StandardScaler().fit_transform(df[features]),
                 columns=features, index=df.index)
X = sm.add_constant(X)

# ------------------ 2. MODELO ------------------
modelo = sm.Logit(y, X).fit(disp=False)
print(modelo.summary())

# Odds ratios (por +1 desviación estándar)
ci = modelo.conf_int()
or_tab = pd.DataFrame({'OR': np.exp(modelo.params),
                       'IC95_inf': np.exp(ci[0]),
                       'IC95_sup': np.exp(ci[1]),
                       'p_value': modelo.pvalues}).drop('const')
print("\nOdds ratios:\n", or_tab.round(4))
print(f"\nPseudo-R² McFadden: {modelo.prsquared:.3f} | "
      f"AUC: {roc_auc_score(y, modelo.predict(X)):.3f}")

# Multicolinealidad (VIF > 5 es preocupante)
vif = pd.Series([variance_inflation_factor(X.values, i) for i in range(1, X.shape[1])],
                index=features)
print("\nVIF:\n", vif.round(2))
```

Tanto el coeficiente de Pseudo-R² (<0.2) como el valor de AUC (0.6) son bajos, lo que significa que este modelo no tiene casi poder predictivo, es decir, las variables independientes no explican casi nada de la variable dependiente.  
El p-valor de sugars_100g y saturated_fat_100g indican que sí tienen una relación con los valores de nova_group, pero viendo los valores de R² y AUC, podemos decir que la asociación existe pero su magnitud es muy pequeña.

#### 4.2 - Análisis de la hipótesis

Para corroborar la hipótesis planteada, utilizamos el test de hipótesis para una proporción poblacional (Prueba Z de una proporción).

```python
# Crear columna altoen_grasas: es 1 si es alto en grasas saturadas

h4_dataset = dataset[dataset["saturated_fat_100g"].notna()]
h4_dataset['altoen_grasas'] = (h4_dataset['saturated_fat_100g'] > 5).astype(int)

# Métricas de la muestra
n = h4_dataset['altoen_grasas'].count()         # Número total de observaciones
x = h4_dataset['altoen_grasas'].sum()           # Cantidad con altoen_grasas = 1
p_muest = x / n                                 # Proporción muestral
p_hip = 0.25                                    # Valor de la hipótesis nula

# Ejecutar la prueba de hipótesis para proporciones
z_stat, p_value = proportions_ztest(count=x, nobs=n, value=p_hip, alternative='larger')

# Mostrar resultados
print("=== RESULTADOS DEL TEST DE HIPÓTESIS ===")
print(f"Tamaño muestral (n)            : {n}")
print(f"Productos altos en grasas saturadas  : {x}")
print(f"Proporción observada (p̂)       : {p_muest:.4f} ({p_muest*100:.2f}%)")
print(f"Estadístico Z de prueba        : {z_stat:.4f}")
print(f"p-valor (p-value)              : {p_value:.4f}")

# Conclusión estadística
alpha = 0.05
if p_value < alpha:
    print(f"\nDecisión: Rechazar H0 (p-valor < {alpha}).")
    print("Conclusión: Existe evidencia suficiente para afirmar que más del 25% de los productos son altos en grasas saturadas.")
else:
    print(f"\nDecisión: No rechazar H0 (p-valor >= {alpha}).")
    print("Conclusión: No existe evidencia suficiente para afirmar que más del 25% de los productos son altos en grasas saturadas.")
```

```python
plant_based = tiene("en:plant-based-foods-and-beverages")
snack = tiene("en:snacks")
lacteo = tiene("en:dairies")
snack_dulce = tiene("en:sweet-snacks")
bebida = tiene("en:beverages")
desayuno = tiene("en:breakfasts")
comida_fermentada = tiene("en:fermented-foods")
cereales_y_papas = tiene("en:cereals-and-potatoes")
condimentos = tiene("en:condiments")

categorias = {
    "plant_based":          plant_based,
    "snack":     snack,
    "lacteo":     lacteo,
    "snack_dulce":     snack_dulce,
    "bebida":          bebida,
    "desayuno":     desayuno,
    "comida_fermentada":     comida_fermentada,
    "cereales_y_papas":     cereales_y_papas,  
    "condimentos":     condimentos,
    "otras":     otras,  
}

dataset2_h6 = dataset_h6.drop(columns=["energy_kcal","fat_100g", "saturated_fat_100g", "carbs_100g", "sugars_100g", "fiber_100g", "proteins_100g", "salt_100g","sodium_100g"])
for nombre, cat in categorias.items():
    dataset2_h6[nombre] = cat
    dataset2_h6[nombre] = dataset2_h6[nombre].apply(lambda x: 1 if x else 0)
dataset2_h6.head(5)
    
```

- Se queda con los registros donde nutriscore_grade no es null ni unknown.
- Creamos variable binaria 'nutriscore_desfavorable', que será 1 si el nutriscore del producto es D o E. 

```python
dataset2_h6["cant_alergenos"] = dataset2_h6["allergens"].apply(
    lambda x: 0 if pd.isna(x) else len([a for a in x.split(",") if a.strip()])
)

dataset2_h6 = dataset2_h6[dataset2_h6["nova_group"].notna()]
dataset2_h6 = dataset2_h6[dataset2_h6["nutriscore_grade"].notna() & ~dataset2_h6["nutriscore_grade"].isin(["UNKNOWN"])]
dataset2_h6["nutriscore_desfavorable"] = dataset2_h6["nutriscore_grade"].apply(lambda x: 1 if x in ["D","E"] else 0)
```

#### 6.2 - Análisis de la hipótesis

Realizamos la regresión logística: Separamos las variables dependientes con las independientes y particionamos el conjunto de entrenamiento con el conjunto de prueba

```python
from sklearn.linear_model import LogisticRegression
import statsmodels.api as sm
from sklearn.model_selection import train_test_split

variables_x = ["cant_alergenos","nova_group"]
for nom , _ in categorias.items():
    variables_x.append(nom)


independientes = dataset2_h6[variables_x]

dependiente = dataset2_h6["nutriscore_desfavorable"]

X_train, X_test, y_train, y_test = train_test_split(independientes, dependiente, test_size=0.2, stratify=dependiente, random_state=1012)
```

Vamos a comprobar que la división se haya hecho de tal manera que el conjunto de tests y el conjunto de entrenamiento tengan una distribución similar respecto a nuestra variable objetivo.

```python
import matplotlib.pyplot as plt

train_class_counts = pd.Series(y_train).value_counts()
test_class_counts = pd.Series(y_test).value_counts()


fig, axes = plt.subplots(1, 2, figsize=(12, 5))


axes[0].bar(train_class_counts.index, train_class_counts.values)
axes[0].set_title('Cantidad de muestras por clase (Conjunto de entrenamiento)')
axes[0].set_xlabel('Clase')
axes[0].set_ylabel('Cantidad de muestras')
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(["No desfavorable", "desfavorable"])

axes[1].bar(test_class_counts.index, test_class_counts.values)
axes[1].set_title('Cantidad de muestras por clase (Conjunto de test)')
axes[1].set_xlabel('Clase')
axes[1].set_ylabel('Cantidad de muestras')
axes[1].set_xticks([0,1])
axes[1].set_xticklabels(["No desfavorable", "desfavorable"])

plt.tight_layout()
plt.show()
```

Bien, vemos que las distribuciones son similares. Vamos a estandarizar los datos de entrenamiento

```python
scaler = StandardScaler()
scaler.fit(X_train)
X_train_stand = scaler.transform(X_train)
X_test_stand = scaler.transform(X_test)
```

A continuación, vamos a probar la exactitud del modelo sin regulizador y con regularizador, probando con diferentes valores de alpha.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import cross_val_score
import warnings


param_grid = {'C': [0.001, 0.01, 0.05, 0.1 ,0.5, 1, 2, 10, 100]}

# Crear un modelo de Regresión Logística
logistic_model = LogisticRegression(penalty=None, max_iter=1000, random_state=10124)
logistic_model_regularizador = LogisticRegression(penalty="l2", max_iter=1000, random_state=10124)

grid_search = GridSearchCV(logistic_model_regularizador, param_grid, cv=5, scoring='accuracy')


with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    grid_search.fit(X_train_stand, y_train)
    logistic_model.fit(X_train_stand, y_train)


print("Modelo sin regularizador")
print("Exactitud:", logistic_model.score(X_train_stand, y_train),"\n")

print("Modelo con regularizador")
print("Mejores parámetros:", grid_search.best_params_)
print("Mejor exactitud:", grid_search.best_score_)

```

Vemos que usando regularizador con alpha=0.01, el modelo mejora levemente, por lo que vamos a quedarnos con este para hacer los tests.

```python
best_logistic_model = grid_search.best_estimator_
theta_logistic = best_logistic_model.coef_

fig, ax = plt.subplots(figsize=(12, 6))
ax.bar(np.arange(0, theta_logistic.size), np.squeeze(theta_logistic))
labels = independientes.columns
ax.set_xticks(np.arange(0, theta_logistic.size)) 
ax.set_xticklabels(labels=labels, rotation=90)
ax.set_title('Pesos aprendidos por el regresor logístico')
plt.show()
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

y_pred_original = best_logistic_model.predict(X_test_stand)

labels = ["No desfavorable", "Desfavorable"]

accuracy_original = accuracy_score(y_test, y_pred_original)
confusion_matrix_original = confusion_matrix(y_test, y_pred_original)
sns.heatmap(confusion_matrix_original, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicho')
plt.ylabel('Valor real')
plt.title('Matriz de confusión')
plt.show()
```
